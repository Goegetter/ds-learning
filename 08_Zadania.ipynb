```python
def splitting(dane,k):
    fold_size = len(dane)//k
    folds = []

    folds_left = k
    y = 0
    while folds_left > 0:
        x = y + fold_size
        if folds_left == 1:
            x += len(dane) % k
        folds.append(dane[y:x])
        y += fold_size
        folds_left -= 1
    return folds

def k_fold_split(dane, k):
    if len(dane) % k == 0:
        return splitting(dane, k)
    else:
        rest = dane[(-1)*(len(dane) % k):]
        return list(splitting(dane, k) + rest)


dane = list(range(1,101))
k = 5
print(k_fold_split(dane, k))


```
