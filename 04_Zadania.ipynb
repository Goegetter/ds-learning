**Zadanie 11**

```python
learning_rate = 0.01
batch_size_11 = 64
batch_size_11_lg = math.log(batch_size_11)
epochs = 100
dropout = 0.5

if learning_rate >0 and learning_rate <= 1:
    print('learning_rate: OK')
else:
    print(f'Learning_rate not in range (0,1]: {learning_rate}')

if batch_size_11_lg == 2:
    print('batch_size_11: OK')
else:
    print(f'batch_size_11 is not exponentiation o 2: {batch_size_11_lg:.2f}')

if epochs >= 0 & epochs <= 1000:
    print('epochs: OK')
else:
    print(f'epochs not in range [0,1000]: {epochs}')

if dropout > 0 and dropout < 1:
    print('dropout: OK')
else:
    print(f'dropout not in range [0,1]: {dropout}')
```

**Zadanie 12**

```python
samples = 50000
batch_size_12 = 128
epochs = 50
time_per_batch_s = 0.15

full_batches = samples // batch_size_12
remain_samples = samples % batch_size_12
print(f'Full batches per epoch: {full_batches}\nSamples in batch remain: {remain_samples}')

#Jest tylko 1 batch niepełny z 80 próbkami
complete_time_s = time_per_batch_s * (full_batches + 1) * epochs
print(f'Total time: {complete_time_s:.1f}s\n')

complete_time_m = complete_time_s / 60
complete_time_h = complete_time_s / 3600

print(f'Estimated time: {complete_time_s // 3600:.0f}h {complete_time_m:.0f}m {complete_time_h:.0f}s')
```

**Zadanie 13**

```python
BP_mmHg = [120, 125, 118, 190, 122]
BP_mmHg_avg = sum(BP_mmHg)/len(BP_mmHg)
BP_mmHg_dev = [(value - BP_mmHg_avg)**2 for value in BP_mmHg]
BP_mmHg_st_dev = (sum(BP_mmHg_dev)/(len(BP_mmHg_dev)-1))**(1/2)

print(f'Average blood pressure: {BP_mmHg_avg:.0f}\nStandard deviation of blood pressure: {BP_mmHg_st_dev:.2f}')

# :C
for value in BP_mmHg:
    if abs(value - BP_mmHg_avg) > 3 * BP_mmHg_st_dev:
        print(f'Wartość: {value} - ANOMALIA!')
    else:
        print(f'Wartość: {value} - OK')
```
