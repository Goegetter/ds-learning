**Zadanie 15**

```python
logs = [
"2024-01-15 10:23:45 INFO Model training started",
"2024-01-15 10:45:12 WARNING Low memory",
"2024-01-15 11:02:33 ERROR Training failed",
"2024-01-15 11:05:01 INFO Retrying with smaller batch"
]

all_err = []
for log in logs:
    value = log.strip().split(' ')
    all_err.append(value[2])

unique_err = []
for unique in all_err:
    if unique not in unique_err:
        unique_err.append(unique)

count_dict = {}
for err in unique_err:
    count_dict[err] = all_err.count(err)
print(count_dict)

for err, count in count_dict.items():
    percent = count / len(all_err)
    print(f'{count_dict[err]}: {percent:.0%}')
```
