```python
from numpy.ma.extras import average


class BankAccount:

    def __init__(self, wlasciciel: str, saldo= 0):
        self.wlasciciel = wlasciciel
        self.saldo = saldo

    def wplac(self, kwota):
        self.saldo += kwota

    def wyplac(self, kwota):
        self.saldo -= kwota

    def pokaz_saldo(self):
        print(f'Wlasciciel: {self.wlasciciel}, saldo: {self.saldo}')

konto_1 = BankAccount('Janek')
konto_1.wplac(1000)
konto_1.wyplac(200)
konto_1.pokaz_saldo()

konto_2 = BankAccount('Piotr')
konto_2.pokaz_saldo()
konto_2.wplac(0.7)
konto_2.pokaz_saldo()
```

```python
class Student:

    def __init__(self, imie: str, numer_indeksu: str, oceny: list[int] | None = None):
        self.imie = imie
        self.numer_indeksu = numer_indeksu
        self.oceny = oceny if oceny is not None else []

    def dodaj_ocene(self, ocena):
        self.oceny.append(ocena)

    def srednia(self):
        return sum(self.oceny)/len(self.oceny)

    def __str__(self):
        # return f'Imie: {self.imie}\nNumer Indeksu: {self.numer_indeksu}\nOcena: {self.oceny}\nŚrednia: {self.srednia()}'
        return f'Imie: {self.imie}\nNumer Indeksu: {self.numer_indeksu}\nOcena: {self.oceny}'


student_1 = Student('Sebastian', '01')
student_1.dodaj_ocene(5)
student_1.dodaj_ocene(4)
print(str(student_1))
print(f'Średnia: {student_1.srednia()}')

student_2 = Student('Agnieszka', '02')
print(str(student_2))
```

```python
class Uczen:

    def __init__(self, imie: str, oceny: list[int] | None = None):
        self.imie = imie
        self.oceny = oceny if oceny is not None else []


class Klasa:

    def __init__(self, nazwa: str, uczniowie):
        self.nazwa = nazwa
        self.uczniowie = uczniowie

    def dodaj_ucznia(self, uczen):
        self.uczniowie.append(uczen)

    def srednia_klasy(self):
        lista_wszystkich_ocen = []
        for uczen in self.uczniowie:
            lista_wszystkich_ocen.extend(uczen.oceny)
        return sum(lista_wszystkich_ocen)/len(lista_wszystkich_ocen)

    def najlepszy_uczen(self):
        srednia_ucznia = {}
        for uczen in self.uczniowie:
            srednia_ucznia[uczen.imie] = sum(uczen.oceny)/len(uczen.oceny)
        return max(srednia_ucznia, key=srednia_ucznia.get)


zbior_ucziow = {'Rafał': [5,4,3], 'Łukasz':[5,3], 'Natalia': [6,3,4], 'Asia': [5,5,3,6], 'Andrzej': [3,3]}

klasa_1 = Klasa('3C', [])

for key, value in zbior_ucziow.items():
    uczen_rekord = Uczen(key, value)
    klasa_1.dodaj_ucznia(uczen_rekord)

print(f'Klasa: {klasa_1.nazwa}\nŚrednia klasy: {klasa_1.srednia_klasy():.2f}\nNajlepszy uczeń: {klasa_1.najlepszy_uczen()}')
```
