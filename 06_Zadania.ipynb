**Zadanie 14**

```python
from unicodedata import category

studenci = [
('Anna', 'Kowalska', 5.0, 4.5, 5.0),
('Jan', 'Nowak', 3.0, 3.5, 3.0),
('Maria', 'Zielińska', 4.0, 4.5, 5.0),
('Piotr', 'Wiśniewski', 2.0, 3.0, 2.5),
('Ewa', 'Wójcik', 5.0, 5.0, 4.5),
('Tomasz', 'Kamiński', 4.0, 3.5, 4.0)
]
```

```python
studenci_mean = []
for line in studenci:
    mean = sum(line[2:])/(len(line)-2)
    studenci_mean.append((*line[:2],round(mean,1)))

category = {}
for student in studenci_mean:
    if student[2] >= 4.5:
        category.setdefault('wyróżniający',[]).append(student)
    elif student[2] >= 3.5 and student[2] < 4.5:
        category.setdefault('dobry',[]).append(student)
    else:
        category.setdefault('średni',[]).append(student)

print(category)
```

**Zadanie 15**

```python
dzien1 = {'user001', 'user002', 'user003', 'user004', 'user005', 'user010'}
dzien2 = {'user002', 'user003', 'user006', 'user007', 'user008', 'user010'}
dzien3 = {'user003', 'user004', 'user009', 'user010', 'user011'}
```

```python
active_3_days = dzien1 & dzien2 & dzien3

active_1_day = {
'day1': dzien1 - dzien2 - dzien3,
'day2': dzien2 - dzien3 - dzien1,
'day3': dzien3 - dzien1 - dzien2,
}

active_2_days = ((dzien1 & dzien2) | (dzien2 & dzien3) | (dzien1 & dzien3)) - active_3_days

print(f'Users active in all of 3 days: {active_3_days}')
print(f'Users active single day: {active_1_day}')
print(f'Users active exactly 2 days: {active_2_days}')
```

```python
filmy_gatunki = {
'Matrix': ['Akcja', 'Sci-Fi'],
'Shrek': ['Komedia', 'Familijny', 'Animacja'],
'Inception': ['Akcja', 'Sci-Fi', 'Thriller'],
'Toy Story': ['Animacja', 'Familijny'],
'Die Hard': ['Akcja', 'Thriller'],
'The Hangover': ['Komedia']
}
```

```python
gatunki_filmy = {}
for filmy, gatunki in filmy_gatunki.items():
    for gatunek in gatunki:
        gatunki_filmy.setdefault(gatunek, set()).add(filmy)

print(gatunki_filmy)
```
